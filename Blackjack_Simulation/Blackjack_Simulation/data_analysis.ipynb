```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("blackjack_results.csv")
num_tests = len(data)
# display(data)
win_prob = data.iloc[:, 0]
lose_prob = data.iloc[:, 1]
tie_prob = data.iloc[:, 2]
```

```python
plt.hist(win_prob, edgecolor='k', color="teal", alpha=0.65)
plt.title("Histogram of win probabilities")
plt.ylabel("Frequency")
plt.xlabel("Probability")
plt.show()
print(f"Mean: {np.mean(win_prob):.6g}")
print(f"Standard Deviation: {np.std(win_prob):.6g}")
print(f"Standard Error: {np.std(win_prob)/np.sqrt(num_tests):.6g}")
```

```python
plt.hist(lose_prob, edgecolor='k', color="red", alpha=0.65)
plt.title("Histogram of lose probabilities")
plt.ylabel("Frequency")
plt.xlabel("Probability")
plt.show()
print(f"Mean: {np.mean(lose_prob):.6g}")
print(f"Standard Deviation: {np.std(lose_prob):.6g}")
print(f"Standard Error: {np.std(lose_prob)/np.sqrt(num_tests):.6g}")
```

```python
plt.hist(tie_prob, edgecolor='k', color="green", alpha=0.65)
plt.title("Histogram of tie probabilities")
plt.ylabel("Frequency")
plt.xlabel("Probability")
plt.show()
print(f"Mean: {np.mean(tie_prob):.6g}")
print(f"Standard Deviation: {np.std(tie_prob):.6g}")
print(f"Standard Error: {np.std(tie_prob)/np.sqrt(num_tests):.6g}")
```

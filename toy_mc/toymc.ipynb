```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("toymc_data.csv")
display(df)
```

```python
transmitted = np.array(df.iloc[:, 0])
absorbed = np.array(df.iloc[:, 1])
fissioned = np.array(df.iloc[:, 2])
# print(transmitted[0:5])
```

```python
plt.hist(transmitted, edgecolor='k', color="teal", alpha=0.75)
plt.title("Histogram of transmitted particles")
plt.ylabel("Frequency")
plt.xlabel("Probability")
plt.show()
print(f"Mean: {np.mean(transmitted):.4g}")
print(f"Standard Deviation: {np.std(transmitted):.4g}")
```

```python
plt.hist(absorbed, edgecolor='k', color="purple", alpha=0.75)
plt.title("Histogram of absorbed particles")
plt.ylabel("Frequency")
plt.xlabel("Probability")
plt.show()
print(f"Mean: {np.mean(absorbed):.4g}")
print(f"Standard Deviation: {np.std(absorbed):.4g}")
```

```python
plt.hist(fissioned, edgecolor='k', color="gold", alpha=0.6)
plt.title("Histogram of fissioned particles")
plt.ylabel("Frequency")
plt.xlabel("Probability")
plt.show()
print(f"Mean: {np.mean(fissioned):.4g}")
print(f"Standard Deviation: {np.std(fissioned):.4g}")
```
